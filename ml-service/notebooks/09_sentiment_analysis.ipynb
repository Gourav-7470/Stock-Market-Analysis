# 09 Financial News NLP Sentiment Analysis

Financial headline text cleaning, VADER/FinBERT sentiment scoring, positive/neutral/negative breakdown.

### Analyze News Headlines with VADER NLP

```python
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
nltk.download('vader_lexicon', quiet=True)

sia = SentimentIntensityAnalyzer()
headlines = [
    'Reliance Industries reports stellar quarter with record EBITDA growth',
    'Global market uncertainty weighs down technology stocks',
    'Central bank keeps interest rates unchanged in monetary policy review'
]

for h in headlines:
    score = sia.polarity_scores(h)
    print(f'Headline: {h}\nScore: {score}\n')
```
