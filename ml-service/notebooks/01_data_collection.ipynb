# 01 Data Collection

Module for downloading historical OHLCV stock price data from reliable APIs (yfinance / NSE).

### Fetch Historical Stock Data using yfinance

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

symbols = ['RELIANCE.NS', 'TCS.NS', 'INFY.NS', 'HDFCBANK.NS']
period = '2y'

data = {}
for sym in symbols:
    ticker = yf.Ticker(sym)
    df = ticker.history(period=period)
    data[sym] = df
    print(f'Fetched {len(df)} records for {sym}')
```

### Inspect Raw OHLCV DataFrame Structure

```python
df_reliance = data['RELIANCE.NS']
print(df_reliance.head())
print(df_reliance.tail())
```
