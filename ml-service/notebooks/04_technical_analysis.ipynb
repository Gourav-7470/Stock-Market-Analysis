# 04 Technical Indicator Engine

Calculate Moving Averages (SMA, EMA), RSI, MACD, Stochastic Oscillator, Bollinger Bands, ATR, OBV.

### Calculate Technical Indicators

```python
df_clean['SMA_20'] = df_clean['Close'].rolling(20).mean()
df_clean['SMA_50'] = df_clean['Close'].rolling(50).mean()
df_clean['EMA_20'] = df_clean['Close'].ewm(span=20).mean()

# RSI Calculation
delta = df_clean['Close'].diff()
gain = (delta.where(delta > 0, 0)).rolling(14).mean()
loss = (-delta.where(delta < 0, 0)).rolling(14).mean()
rs = gain / loss
df_clean['RSI_14'] = 100 - (100 / (1 + rs))

print('Technical Indicators calculated successfully!')
```
