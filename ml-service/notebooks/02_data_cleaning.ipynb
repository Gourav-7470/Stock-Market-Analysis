### Clean Missing Data & Handle Outliers

```python
import pandas as pd
import numpy as np

# Load raw data
df = df_reliance.copy()

# Missing value analysis
print('Missing values per column:')
print(df.isnull().sum())

# Forward fill & backward fill missing prices
df_clean = df.ffill().bfill()

# Outlier analysis using Z-Score
z_scores = np.abs((df_clean['Close'] - df_clean['Close'].mean()) / df_clean['Close'].std())
outliers = df_clean[z_scores > 3]
print(f'Detected {len(outliers)} statistical outliers.')
```
