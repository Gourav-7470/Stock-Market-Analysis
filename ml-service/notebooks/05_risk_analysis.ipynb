# 05 Volatility & Risk Analysis

Compute Daily Volatility, Annualized Volatility, Maximum Drawdown, Historical 95% VaR, Sharpe Ratio.

### Compute Risk & Volatility Metrics

```python
daily_vol = df_clean['Daily_Return'].std()
annualized_vol = daily_vol * np.sqrt(252)

# Peak & Drawdown
peak = df_clean['Close'].cummax()
drawdown = (df_clean['Close'] - peak) / peak
max_drawdown = drawdown.min()

# Historical 95% VaR
var_95 = np.percentile(df_clean['Daily_Return'].dropna(), 5)

print(f'Annualized Volatility: {annualized_vol*100:.2f}%')
print(f'Maximum Drawdown: {max_drawdown*100:.2f}%')
print(f'Historical 95% VaR (1-day): {var_95*100:.2f}%')
```
