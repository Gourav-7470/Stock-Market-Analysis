### Return Distributions & Cumulative Performance Growth

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate returns
df_clean['Daily_Return'] = df_clean['Close'].pct_change()
df_clean['Cumulative_Return'] = (1 + df_clean['Daily_Return']).cumprod() - 1

plt.figure(figsize=(12, 5))
plt.plot(df_clean.index, df_clean['Cumulative_Return'] * 100, label='Cumulative Return %', color='green')
plt.title('Reliance Industries Cumulative Return (%)')
plt.xlabel('Date')
plt.ylabel('Return (%)')
plt.grid(True)
plt.show()
```
