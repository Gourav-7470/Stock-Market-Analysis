Engineers lag features, momentum indicators, rolling volatility, volume shifts for ML models.

### Create Lag Features & Targets

```python
for lag in [1, 2, 5, 10]:
    df_clean[f'Lag_{lag}'] = df_clean['Close'].shift(lag)

df_clean['Target_5D'] = df_clean['Close'].shift(-5)
feature_df = df_clean.dropna()
print(f'Feature Matrix shape: {feature_df.shape}')
```
